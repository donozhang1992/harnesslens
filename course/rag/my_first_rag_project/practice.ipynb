```python
from langchain_text_splitters import RecursiveJsonSplitter
import tiktoken

encoding = tiktoken.get_encoding("o200k_base")

json_data = {
    "user": {
        "id": 123,
        "name": "Alice",
        "profile": {
            "bio": "Alice is a data scientist who loves working with large language models.",
            "location": "Stockholm",
            "location1": "Stockholm",
            "location2": "Stockholm",
            "location3": "Stockholm",
            "location4": "Stockholm",
            "location5": "Stockholm",
            "interests": ["AI", "ML", "NLP", "Data Engineering"],
        },
    },
    "posts": [
        {
            "id": 1,
            "title": "Getting started with LangChain",
            "content": "LangChain makes it easier to build LLM-powered applications...",
        },
        {
            "id": 2,
            "title": "Recursive JSON splitting",
            "content": "RecursiveJsonSplitter can break nested JSON into manageable chunks.",
        },
    ],
}

splitter = RecursiveJsonSplitter(max_chunk_size=20)
json_chunks = splitter.split_json(json_data)
for i, chunk in enumerate(json_chunks):
    print(f"Chunk {i+1}: {chunk}")
    print(f"Token count: {len(encoding.encode(str(chunk)))}")
```

```python
from langchain_community.document_loaders import JSONLoader


file_path = "data/common.jsonl"
loader = JSONLoader(
    file_path = file_path,
    json_lines=True,
    jq_schema=".",
    text_content=False,
    is_content_key_jq_parsable=True,
    content_key=".question",
    metadata_func=lambda data, meta: {
        **meta,
        "topic": data["topic"],
        "answer": data["answer"],
    },

)

documents = loader.load()
print(len(documents))
for doc in documents:
    print("page_content:", doc.page_content)
    print("metadata:", doc.metadata)
    print("-----")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings


model_name = "Snowflake/snowflake-arctic-embed-l-v2.0"
# model_name = "sentence-transformers/all-MiniLM-L6-v2"
embedder = HuggingFaceEmbeddings(model_name=model_name)
```

```python
from langchain_chroma import Chroma


vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embedder,
    # persist_directory="data/chroma_db",
    collection_name="my_collection",
    collection_metadata={"hnsw:space": "cosine"},
)

# retrieval
query = "How many days of vacation can we get each year?"
docs_and_scores = vectorstore.similarity_search_with_score(
    query, 
    k=5,
    # filter={"topic": "Human Resources"},
)

for doc, score in docs_and_scores:
    print("Score:", score)
    print("Content:")
    print(doc.page_content)
    print(doc.metadata['answer'])
    print("Topic:", doc.metadata["topic"])
    print('----------')
```
