````python
from langchain_chroma import Chroma
import chromadb
import shutil
import os

# Clear previously persisted data to prevent duplicate insertions on re-run
chroma_path = "data/chroma_db"
if os.path.exists(chroma_path):
    shutil.rmtree(chroma_path)

vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embedder,
    persist_directory=chroma_path,
    collection_name="my_collection",
    collection_metadata={"hnsw:space": "cosine"},
)

# Chroma reports cosine *distance* in [0, 2]; convert to similarity in [0, 1]
class CosineSimilarityCollection(chromadb.api.models.Collection.Collection):
    def query(self, *args, **kwargs):
        res = super().query(*args, **kwargs)
        res['distances'] = [[1 - d for d in row] for row in res['distances']]
        return res
vectorstore._collection.__class__ = CosineSimilarityCollection

# retrieval
query = "How many days of vacation can we get each year?"
docs_and_scores = vectorstore.similarity_search_with_score(
    query, 
    k=5,
    filter={"topic": "Human Resources"},
)

for doc, score in docs_and_scores:
    print("Score:", score)
    print("Content:")
    print(doc.page_content)
    print(doc.metadata['answer'])
    print("Topic:", doc.metadata["topic"])
    print('----------')
````

## RAG based on Chain

````python

from langchain_community.vectorstores import FAISS

def custom_relevance_score_fn(distance: float) -> float:
    return distance

vectore_db = FAISS.from_documents(
    documents,
    embedding=embedder,
    distance_strategy="MAX_INNER_PRODUCT",
    relevance_score_fn=custom_relevance_score_fn
)

def search_filter(metadata):
    return metadata["topic"]=="Human Resources"
retriever = vectore_db.as_retriever(
    # search_type: similarity/similarity_score_threshold/mmr 
    search_type = "similarity_score_threshold",
    search_kwargs = {
        "k": 5,
        "score_threshold": 0.6,
        "fetch_k": 20,
        # "filter": search_filter
    }
)
# query = "How should I handle a customer complaint?"
query = "How many days of vacation can we get each year?"
retriever.invoke(query)
````

````python
def docs_to_context(docs):
    context = "\n\n".join(f"{doc.page_content}\n{doc.metadata['answer']}" for doc in docs)
    print(docs)
    print('--------')
    print(context)
    print('--------')
    return {
        "docs": docs,
        "context": context
    }
````

````python
from langchain_core.prompts import PromptTemplate

prompt_template = """\
You are a helpful company internal assistant.
Answer the question using ONLY the context below.
You can paraphrase and infer when wording is different but meaning is the same.
If the context truly does not provide enough information, say "I don't know".

Context:
{context}

Question:
{question}
"""


prompt = PromptTemplate(
    input_variables=["context", "question"],
    template=prompt_template
)
````

````python
import os
from langchain_openai import ChatOpenAI
from langchain_core.runnables import RunnablePassthrough

# DeepSeek is OpenAI-compatible — just point to a different base_url
llm = ChatOpenAI(
    model="deepseek-chat",
    openai_api_key=os.environ["DEEPSEEK_API_KEY"],
    openai_api_base="https://api.deepseek.com",
    temperature=0,
)

rag_chain = (
    {
        "question": RunnablePassthrough(),
        "context": retriever | docs_to_context 
    } | prompt | llm
)
````

## Advanced Retrieval Strategies（高级检索策略）

基础 RAG 仅做"相似度搜索"，但在实际场景中往往不够用。本节介绍三种核心增强技术：

| 策略 | 解决的问题 |
|------|-----------|
| **Multi-Query** | 单一 query 措辞不准，召回率低 |
| **HyDE** | Query 和 Document 语义空间不对齐 |
| **Contextual Compression** | 检索 chunk 含大量无关内容，浪费 context 窗口 |

这三种技术互相正交，可以组合使用。

### 1. Multi-Query Retrieval

**问题**：单一 query 可能措辞不准，导致召回率不足。

**解决**：让 LLM 自动生成多个语义相近的 query 变体，分别检索后去重合并，提升召回率。

```
query → LLM → [q1, q2, q3] → retrieve each → deduplicate → merged results
```

开启 logging 可以看到 LLM 生成了哪些 query 变体。

````python
import logging
from langchain_classic.retrievers.multi_query import MultiQueryRetriever

logging.basicConfig()
logging.getLogger("langchain.retrievers.multi_query").setLevel(logging.INFO)

# MultiQueryRetriever generates N reworded versions of the query and merges results
multi_query_retriever = MultiQueryRetriever.from_llm(
    retriever=vectore_db.as_retriever(search_kwargs={"k": 3}),
    llm=llm,
)

query = "How many days of vacation can we get each year?"
docs = multi_query_retriever.invoke(query)

print(f"\nTotal unique documents retrieved: {len(docs)}")
for doc in docs:
    print(doc.page_content)
    print("-----")
````

### 2. HyDE（Hypothetical Document Embeddings）

**问题**：Query（"vacation policy"）和 Document（"How many annual leave days..."）处于不同的语义空间，导致向量相似度不准。

**解决**：先让 LLM 生成一个假设性答案（hypothetical answer），再用它的向量去搜索——document-to-document 的相似度往往比 query-to-document 更准。

```
query → LLM → hypothetical_answer → embed → search
```

````python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

hyde_prompt = PromptTemplate.from_template(
    "Write a short, plausible answer to the following question:\n\n"
    "Question: {question}\n\n"
    "Hypothetical Answer:"
)

# HyDE chain: question → hypothetical answer → embed & search
hyde_chain = (
    hyde_prompt
    | llm
    | StrOutputParser()
    | vectore_db.as_retriever(search_kwargs={"k": 3})
)

query = "company vacation policy"
docs = hyde_chain.invoke({"question": query})

print(f"HyDE retrieved {len(docs)} documents for query: '{query}'\n")
for doc in docs:
    print(doc.page_content)
    print("-----")
````

### 3. Contextual Compression Retriever

**问题**：检索到的 chunk 可能含有大量与当前问题无关的内容，导致 context 窗口浪费。

**解决**：用 LLM 从每个 chunk 中抽取与 query 相关的部分，去除噪声后再送给生成模型。

````python
from langchain_classic.retrievers.contextual_compression import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors.chain_extract import LLMChainExtractor

# LLMChainExtractor strips irrelevant parts from each retrieved chunk
compressor = LLMChainExtractor.from_llm(llm)

compression_retriever = ContextualCompressionRetriever(
    base_compressor=compressor,
    base_retriever=vectore_db.as_retriever(search_kwargs={"k": 4}),
)

query = "How many days of vacation can we get each year?"
compressed_docs = compression_retriever.invoke(query)

print(f"Retrieved {len(compressed_docs)} compressed documents\n")
for doc in compressed_docs:
    print(doc.page_content)
    print("-----")
````

## Reranking（重排序）

**问题**：向量相似度是 bi-encoder，query 和 document 各自独立编码后比较，精度有限。

**解决**：两阶段检索
1. 向量检索（快）→ 召回 Top-K（如 k=10）
2. Cross-encoder Reranker（慢但精准）→ 精排出 Top-N（如 n=3）

```
query + doc → cross-encoder → relevance score  (考虑了两者的交互)
```

Reranking 显著提升准确率，是生产 RAG 管道的标配步骤。

### FlashRank Reranker（本地，无需 API Key）

FlashRank 是轻量级跨编码器（cross-encoder），在本地运行，不需要 API Key，适合快速实验和隐私敏感场景。

````python
%pip install -q flashrank langchain-community

from langchain_classic.retrievers.contextual_compression import ContextualCompressionRetriever
from langchain_community.document_compressors import FlashrankRerank

# FlashRank: local cross-encoder reranker, no API key needed
reranker = FlashrankRerank(model="ms-marco-MiniLM-L-12-v2", top_n=3)

reranking_retriever = ContextualCompressionRetriever(
    base_compressor=reranker,
    base_retriever=vectore_db.as_retriever(search_kwargs={"k": 10}),  # fetch more, then rerank
)

query = "How many days of vacation can we get each year?"
reranked_docs = reranking_retriever.invoke(query)

for doc in reranked_docs:
    print(f"Relevance: {doc.metadata.get('relevance_score', 'N/A'):.4f}")
    print(f"Content:   {doc.page_content}")
    print("-----")
````

默认 `rag_chain.invoke()` 等待完整结果后一次性返回。生产应用通常需要流式输出（逐 token 显示），提升感知速度。

LangChain LCEL 链原生支持 `.stream()`——无需额外配置，只要链中的 LLM 支持 streaming 即可。

````python
query = "How should I handle a customer complaint?"

print("Streaming response:\n")
for chunk in rag_chain.stream(query):
    print(chunk.content, end="", flush=True)
print()  # final newline
````

## Conversational RAG

实际应用中用户几乎总是会追问上下文相关的问题，例如：

> Q1: How many vacation days do we get?  
> Q2: **And what about medical leave?** ← 如果不知道 Q1，这个问题就无法独立检索

解决方案：**Query Condensation**——先用 LLM 把追问改写为独立问题，再检索。

```
chat_history + follow-up → LLM → standalone question → retriever → answer
```

````python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

# Step 1: Condense a follow-up question into a standalone question
condense_prompt = ChatPromptTemplate.from_messages([
    ("system", (
        "Given the conversation history and the latest user question, "
        "rewrite it as a self-contained question. "
        "If there is no history, return the question unchanged."
    )),
    MessagesPlaceholder("chat_history"),
    ("human", "{question}"),
])
condense_chain = condense_prompt | llm | StrOutputParser()

# Step 2: QA prompt that includes history for conversational tone
qa_prompt = ChatPromptTemplate.from_messages([
    ("system", (
        "You are a helpful company assistant.\n"
        "Answer using ONLY the context below.\n"
        "If the context doesn't have enough information, say 'I don't know'.\n\n"
        "Context:\n{context}"
    )),
    MessagesPlaceholder("chat_history"),
    ("human", "{question}"),
])

def retrieve_with_condense(inputs):
    if inputs.get("chat_history"):
        standalone_q = condense_chain.invoke(inputs)
    else:
        standalone_q = inputs["question"]
    docs = retriever.invoke(standalone_q)
    return "\n\n".join(
        f"{doc.page_content}\n{doc.metadata['answer']}" for doc in docs
    )

conv_rag_chain = (
    RunnablePassthrough.assign(context=retrieve_with_condense)
    | qa_prompt
    | llm
)

# --- Multi-turn demo ---
chat_history = []

q1 = "How many vacation days do we get per year?"
r1 = conv_rag_chain.invoke({"question": q1, "chat_history": chat_history})
print(f"Q: {q1}\nA: {r1.content}\n")
chat_history += [HumanMessage(q1), AIMessage(r1.content)]

q2 = "And what about medical leave?"
r2 = conv_rag_chain.invoke({"question": q2, "chat_history": chat_history})
print(f"Q: {q2}\nA: {r2.content}")
````

## RAG Evaluation (RAGAS)

RAG 的核心挑战之一是**你不知道它在真实场景中有多好用**。RAGAS 提供了一套无需人工标注答案就能评估的指标：

| 指标 | 评估什么 | 是否需要 ground truth |
|------|---------|----------------------|
| **Faithfulness** | 答案是否忠实于检索内容（不幻觉）| 否 |
| **Answer Relevancy** | 答案与问题的相关程度 | 否 |
| **Context Recall** | 相关文档是否被检索到 | 是 |
| **Context Precision** | 检索到的文档中有多少真正相关 | 是 |

````python
%pip install -q ragas datasets
````

````python
from ragas import evaluate
from ragas.metrics import faithfulness, answer_relevancy, context_recall, context_precision
from ragas.llms import LangchainLLMWrapper
from ragas.embeddings import LangchainEmbeddingsWrapper
from datasets import Dataset

# Point RAGAS at DeepSeek instead of OpenAI
ragas_llm = LangchainLLMWrapper(llm)
ragas_embeddings = LangchainEmbeddingsWrapper(embedder)

questions = [
    "How many annual leave days am I entitled to?",
    "What should I do if a customer is dissatisfied with the service?",
    "How does the performance review process work?",
]
ground_truths = [
    "Typically, employees get 10 to 20 days per year.",
    "Apologize for the inconvenience, ask for specifics, and work to resolve the issue.",
    "Performance reviews are usually conducted annually or semi-annually.",
]

answers, contexts = [], []
for q in questions:
    docs = retriever.invoke(q)
    answer = rag_chain.invoke(q)
    contexts.append([doc.page_content + "\n" + doc.metadata["answer"] for doc in docs])
    answers.append(answer.content)

eval_dataset = Dataset.from_dict({
    "question": questions,
    "answer": answers,
    "contexts": contexts,
    "ground_truth": ground_truths,
})

result = evaluate(
    eval_dataset,
    metrics=[faithfulness, answer_relevancy, context_recall, context_precision],
    llm=ragas_llm,
    embeddings=ragas_embeddings,
)
print(result)
result.to_pandas()
````
