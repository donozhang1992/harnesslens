```python
import asyncio
import time

async def ticker():
    for i in range(5):
        print(f"{time.strftime('%X')} ticker {i}")
        await asyncio.sleep(1)

async def good_sleep():
    print(f"{time.strftime('%X')} good_sleep start")
    await asyncio.sleep(5)
    print(f"{time.strftime('%X')} good_sleep end")

await asyncio.gather(
    ticker(),
    good_sleep(),
)
```

```python
import asyncio
import time

async def ticker():
    for i in range(5):
        print(f"{time.strftime('%X')} ticker {i}")
        await asyncio.sleep(1)

async def bad_sleep():
    print(f"{time.strftime('%X')} bad_sleep start")
    time.sleep(5)
    print(f"{time.strftime('%X')} bad_sleep end")

await asyncio.gather(
    ticker(),
    bad_sleep(),
)
```
